```python

array = []
def insert(num):
    global array
    array.append(num)
    print(array)
    i = len(array)-1
    while i>0:
        p = (i-1)//2
        if array[i]<array[p]:
            array[i], array[p] = array[p], array[i]
            i = p
            print(array)
        else:
            break
            

insert(10)
print(array)
insert(20)
print(array)
insert(30)
print(array)
insert(5)
print(array)
```
