```python
from collections import deque

class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None


# Sample Tree
#
#         10
#        /  \
#       5    20
#      / \   / \
#     3   7 15 25
#


class Tree:
    def __init__(self):
        self.root=None

    def insert(self, n):
        new = Node(n)
        if self.root == None:
            self.root = new
            return
            
        temp = self.root
        
        while temp:
            if n < temp.data:
                if temp.left == None:
                    temp.left = new
                    return
                temp = temp.left
                
            else:
                if temp.right == None:
                    temp.right = new
                    return
                temp = temp.right

    def traverse(self, node=None):
        if node == None:
            node = self.root
        if node.left != None:
            self.traverse(node.left)
        print(node.data)
            
        if node.right != None:
            self.traverse(node.right)

    def find_depth(self):
        self.get_depth(self.root, 0)
        
    def get_depth(self, curr, depth):
        # Base condition
        if curr == None:
            print("Depth = ", depth)
            return

        # Call next
        self.get_depth(curr.left, depth+1)
        self.get_depth(curr.right, depth+1)
        
    def print_in_level(self):
        members = [self.root]
        offset = 0
        
        while offset<len(members):
            curr = members[offset]
            offset+=1
            if curr.left != None:
                members.append(curr.left)
            if curr.right != None:
                members.append(curr.right)
        for i in members:
            print(i.data)

t = Tree()
t.insert(10)
t.insert(11)

t.insert(12)
t.insert(13)

t.insert(13)
t.insert(1)
t.insert(0)

t.traverse()
t.find_depth()

t.print_in_level()
```
